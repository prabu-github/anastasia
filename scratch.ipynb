```python
from pathlib import Path
import numpy as np
import pandas as pd
from pprint import pprint
```

```python
orig_dir = Path('origdata')
o_file = orig_dir/'LakeView_hytraitsinput_20260424.csv'
odf = pd.read_csv(o_file)
odf['spectrum_id'] = str(1)
odf['unique_id'] = odf.apply(lambda r: (r['sample_id'] + 
                                        '__' + 
                                        r['spectrum_id']),
                             axis=1)
print(f'{odf.shape = }')
pprint(odf.columns)
```

```python
col_remap = {}
sdf_cols = ['sample_id',
            'spectrum_id',
            'unique_id']
min_wave, max_wave = 3000.0, 0.0

for c in odf.columns:
    try:
        f = float(c)
        col_remap[c] = f'X_{f:4.3f}'
        sdf_cols.append(col_remap[c])
        if f < min_wave:
            min_wave = f
        if f > max_wave:
            max_wave = f
    except:
        col_remap[c] = c

tdf_cols = ['sample_id',
            'TP', 
            'TN', 
            'IC',
            'SRP', 
            'nitrite_nitrate', 
            'ammonium',
            'chl_CF', 
            'temp', 
            'phyco_CF',
            'mmHg', 
            'SPcond', 
            'sulfate',
            'Silica', 
            'DO_pct', 
            'TSS',
            'secchi_avg', 
            'NPOC', 
            'chloride']

odf = odf.rename(columns=col_remap)

sdf = odf[sdf_cols].copy()
tdf = odf[tdf_cols].copy()
```
